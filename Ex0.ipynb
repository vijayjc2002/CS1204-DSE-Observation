## PROCEDURE

### Installation of Anaconda on Ubuntu

Anaconda is the recommended distribution of Python for data science. It comes with Python, JupyterLab and many data science packages (NumPy, Pandas, SciPy, Statsmodels, Matplotlib, Seaborn, Scikit-learn) already installed. In Ubuntu, Anaconda is installed and opened using commands in the **Terminal** (press **Ctrl + Alt + T** to open the Terminal).

**Step 1:** Open the official website **anaconda.com/download** and download the **Linux 64-Bit (x86) Installer**. It is a **.sh** file and is saved in the **Downloads** folder.

**Step 2:** Open the Terminal and run the installer:

```
cd Downloads
bash Anaconda3-*-Linux-x86_64.sh
```

**Step 3:** Answer the prompts in the Terminal:

- Press **Enter** to read the licence, then press **q** to skip to the end.
- Type **yes** to accept the licence.
- Press **Enter** to keep the default folder (`/home/<username>/anaconda3`).
- When it asks to *initialize conda* (update your shell profile), type **yes**.

The installation takes a few minutes.

**Step 4:** Close the Terminal and open it again (or run `source ~/.bashrc`). Check the installation:

```
conda --version
```

`(base)` appears at the start of the Terminal line. This means Anaconda is ready.

**Step 5 (every lab session):** Open the Terminal, go to the lab folder and start JupyterLab:

```
cd Documents/CS1204-DSE-Observation
jupyter lab
```

JupyterLab opens in the browser. **Keep the Terminal open** while working; to stop JupyterLab, press **Ctrl + C** in the Terminal.

_If it shows **jupyter: command not found**, run these two lines:_

```
source ~/anaconda3/bin/activate
jupyter lab
```

_To fix it permanently, run `~/anaconda3/bin/conda init` once, then close and open the Terminal again._

**Step 6:** In JupyterLab, double-click a notebook in the file browser on the left, or click **File → New → Notebook** and select **Python 3**. A notebook has two types of cells:

- **Code cell** — Python code; the output appears just below it.
- **Markdown cell** — text, headings and images to explain the work.



_Markdown and Code cells in a notebook._

**Step 7:** Click a cell and press **Shift + Enter** (or the **Run ▶** button) to run it. Press **Ctrl + S** to save the notebook with its output.

### Features of the packages

- **NumPy** — multidimensional arrays and matrices, with fast mathematical and statistical functions on them. Many other Python libraries are built on NumPy.
- **Pandas** — Series and DataFrame for table-like data; reading CSV/Excel files, sorting, slicing, merging, grouping and handling missing data.
- **SciPy** — algorithms for linear algebra, integration, optimization and statistics; built on NumPy.
- **Statsmodels** — statistical models, tests and estimation (regression, time series); a complement to SciPy.
- **Jupyter (JupyterLab)** — interactive, browser-based platform to write code, text and see output in one notebook.
- **Matplotlib / Seaborn** — plotting and visualization.

## PROGRAM
