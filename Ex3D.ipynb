## AIM

To compare the results of univariate analysis, bivariate analysis and the linear regression model for the UCI Diabetes data set and the Pima Indians Diabetes data set.

## ALGORITHM

**Step 1:** Start the program.

**Step 2:** Import pandas, load_diabetes, LinearRegression and r2_score.

**Step 3:** Load the UCI Diabetes data set (built-in) and read the Pima Indians Diabetes data set from data/diabetes.csv.

**Step 4:** Write a function that finds the skewness and kurtosis of BMI (univariate), the correlation of BMI with the target (bivariate) and the R² score of a linear model.

**Step 5:** Apply the function to both data sets – UCI (target: disease progression) and Pima (target: Glucose).

**Step 7:** Compare the results of the two data sets.

## PROGRAM

_Click the code and press **Shift + Enter** to run it. The output appears just below the code._

```python
import pandas as pd
from sklearn.datasets import load_diabetes
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

# Data set 1: UCI diabetes (built-in), Data set 2: Pima Indians diabetes
uci = load_diabetes(as_frame=True).frame                      # UCI data as a table
pima = pd.read_csv('data/diabetes.csv')

# Univariate, bivariate and regression results for one data set
def analyse(x, y):
    model = LinearRegression().fit(x.to_frame(), y)           # linear model for this data set
    return {'Skewness of BMI': x.skew(),                      # univariate result
            'Kurtosis of BMI': x.kurt(),
            'Correlation (BMI, target)': x.corr(y),           # bivariate result
            'R2 score (linear model)': r2_score(y, model.predict(x.to_frame()))}  # regression result

# Compare the two data sets side by side
result = pd.DataFrame({'UCI (target: progression)': analyse(uci['bmi'], uci['target']),
                       'Pima (target: Glucose)': analyse(pima['BMI'], pima['Glucose'])})
print(result.round(3))                                        # both data sets side by side
```
