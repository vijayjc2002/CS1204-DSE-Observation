## AIM

To apply and explore Density and Contour plotting functions on a two-dimensional function using Matplotlib.

## ALGORITHM

**Step 1:** Start the program.

**Step 2:** Import matplotlib.pyplot and numpy.

**Step 3:** Enable inline plotting using %matplotlib inline.

**Step 4:** Define a function f(x, y) = sin(x)^10 + cos(10+yx)*cos(x).

**Step 5:** Generate x and y arrays using np.linspace.

**Step 6:** Create a 2-D grid using np.meshgrid(x, y) to get X and Y matrices.

**Step 7:** Compute Z = f(X, Y) over the grid.

**Step 8:** Draw a standard contour plot (black lines) using plt.contour().

**Step 9:** Draw a color contour plot using plt.contour() with cmap='RdGy'.

**Step 10:** Draw a filled contour (density) plot using plt.contourf() with colorbar.

**Step 11:** Stop the program.



## PROGRAM

_Click the code and press **Shift + Enter** to run it. The output appears just below the code._

```python
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np

def f(x, y):
    return np.sin(x) ** 10 + np.cos(10 + y * x) * np.cos(x)  # height z for each (x, y)

x = np.linspace(0, 5, 50)
y = np.linspace(0, 5, 40)
X, Y = np.meshgrid(x, y)                                     # 2-D grid from x and y
Z = f(X, Y)                                                  # z value at every grid point

plt.figure(figsize=(6, 5))
plt.contour(X, Y, Z, colors='black')                         # contour = lines of equal height
plt.title('Standard Contour'); plt.xlabel('X'); plt.ylabel('Y')
plt.show()

plt.figure(figsize=(6, 5))
plt.contour(X, Y, Z, 20, cmap='RdGy')                        # 20 levels with a colour map
plt.title('Color Contour'); plt.colorbar(); plt.show()

plt.figure(figsize=(6, 5))
plt.contourf(X, Y, Z, 20, cmap='RdGy')                       # filled contour = density plot
plt.title('Density (Filled Contour)'); plt.colorbar(); plt.show()
```
