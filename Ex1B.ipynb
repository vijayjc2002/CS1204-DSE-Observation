```python
import pandas as pd

# Create a Series
s = pd.Series([10, 20, 30, 40, 50])              # Series = one column with index
print("Series:")
print(s)

# Create a DataFrame
data = {'Name': ['Arun', 'Bala', 'Chitra', 'Divya', 'Esha'],
        'Age' : [20, 21, 22, 20, 23],
        'Marks': [85, 72, 90, 78, 88]}
df = pd.DataFrame(data)                          # DataFrame = table from the dictionary
print("DataFrame:")
print(df)

print("Describe:")
print(df.describe())                             # count, mean, std, min, max ...
print("Names:", df['Name'].tolist())  # one column as a list
print("First 2 rows:")
print(df.head(2))                                # first n rows
print("Sorted by Marks:")
print(df.sort_values('Marks', ascending=False))  # sort rows by a column
print("Mean marks by Age:")
print(df.groupby('Age')['Marks'].mean())         # group by Age, mean of each group
```
