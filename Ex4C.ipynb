_Click the code and press **Shift + Enter** to run it. The output appears just below the code._

```python
%matplotlib inline
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

con = pd.read_csv('data/diabetes.csv')

plt.figure(figsize=(7, 5))
sns.scatterplot(x="Pregnancies", y="Age", data=con)           # one dot per patient
plt.title('Scatter: Pregnancies vs Age'); plt.show()

sns.lmplot(x="Pregnancies", y="Age", data=con, height=4, aspect=1.4)  # scatter with regression line
plt.title('Regression Scatter'); plt.show()

r = stats.pearsonr(con['Age'], con['Outcome'])                # Pearson r between -1 and 1
print("Pearson coefficient (Age vs Outcome):", r)

cormat = con.corr()                                           # correlation of every pair of columns
plt.figure(figsize=(8, 6))
sns.heatmap(round(cormat, 2), annot=True, cmap='coolwarm')    # colour box for each correlation
plt.title('Correlation Heatmap'); plt.show()
```
