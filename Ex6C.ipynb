## AIM

To build a cartographic (map-based) visualization of a dataset involving districts in India using Python and Plotly.

## ALGORITHM

**Step 1:** Start the program.

**Step 2:** Import pandas, plotly.express and plotly.io.

**Step 3:** Set the Plotly renderer to 'notebook' for inline display.

**Step 4:** Load the Asia map outline from the data folder, so no internet is needed.

**Step 5:** Create a DataFrame of the rainfall of ten districts with their latitude and longitude.

**Step 6:** Draw a scatter_geo map of district rainfall using px.scatter_geo().

**Step 7:** Show the Asia region with country borders, set the size and display the map.

**Step 8:** Stop the program.

## PROGRAM

_Click the code and press **Shift + Enter** to run it. The output appears just below the code._

```python
%matplotlib inline
import pandas as pd
import plotly.express as px
import plotly.io as pio
from IPython.display import display, HTML
pio.renderers.default = "notebook"

# Load the map outline from the data folder (so no internet is needed)
asia_topo = open('data/asia_110m.json').read()
display(HTML("<script>window.PlotlyGeoAssets = window.PlotlyGeoAssets || {topojson: {}};" +
             "PlotlyGeoAssets.topojson['asia_110m'] = " + asia_topo + ";" + "</script>"))

district_data = pd.DataFrame({
    'District' : ['Chennai','Mumbai','Kolkata','Delhi','Bangalore',
                  'Hyderabad','Ahmedabad','Pune','Jaipur','Lucknow'],
    'Latitude' : [13.08,19.08,22.57,28.70,12.97,17.38,23.03,18.52,26.91,26.85],
    'Longitude': [80.27,72.88,88.36,77.10,77.59,78.49,72.58,73.86,75.79,80.95],
    'Rainfall' : [1400,2200,1600,790,970,810,800,720,650,1010]
})
fig3 = px.scatter_geo(district_data, lat='Latitude', lon='Longitude',  # circle at each district
                      size='Rainfall', color='Rainfall', hover_name='District',  # bigger circle = more rainfall
                      color_continuous_scale='Greens', projection='natural earth',
                      title='India District-wise Rainfall (mm/year)', size_max=30)
fig3.update_geos(scope='asia', showcountries=True)            # show Asia map
fig3.update_layout(width=900, height=650); fig3.show()
```
