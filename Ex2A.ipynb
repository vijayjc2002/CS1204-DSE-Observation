## PROGRAM

_Click the code and press **Shift + Enter** to run it. The output appears just below the code._

```python
# Read the whole file
file1 = open('data/myfile.txt', 'r')  # 'r' = open for reading
print("Output of read():")
print(file1.read())                   # whole file as one string
file1.close()

# Read line by line
file2 = open('data/myfile.txt', 'r')
print("Output of readline():")
print(file2.readline())               # only one line
file2.close()

# Read into list
file3 = open('data/myfile.txt', 'r')
print("Output of readlines():")
print(file3.readlines())              # all lines as a list
file3.close()                         # always close the file
```
