## ALGORITHM

**Step 1:** Start the program.

**Step 2:** Import pandas, plotly.express and plotly.io.

**Step 3:** Set the Plotly renderer to 'notebook' for inline display.

**Step 4:** Load the world map outline from the data folder, so no internet is needed.

**Step 5:** Create a DataFrame of ten countries with their ISO-3 codes and population (in millions).

**Step 6:** Draw a world choropleth map using px.choropleth() with the ISO-3 country codes (locationmode='ISO-3').

**Step 7:** Set the size of the map and display it.

**Step 8:** Stop the program.

## PROGRAM

_Click the code and press **Shift + Enter** to run it. The output appears just below the code._

```python
%matplotlib inline
import pandas as pd
import plotly.express as px
import plotly.io as pio
from IPython.display import display, HTML
pio.renderers.default = "notebook"

# Load the map outline from the data folder (so no internet is needed)
world_topo = open('data/world_110m.json').read()              # world map file (no internet)
display(HTML("<script>window.PlotlyGeoAssets = window.PlotlyGeoAssets || {topojson: {}};" +
             "PlotlyGeoAssets.topojson['world_110m'] = " + world_topo + ";" + "</script>"))

world_data = pd.DataFrame({
    'Country': ['India','China','United States','Indonesia','Pakistan',
                'Brazil','Nigeria','Bangladesh','Russia','Mexico'],
    'Code'   : ['IND','CHN','USA','IDN','PAK','BRA','NGA','BGD','RUS','MEX'],
    'Population': [1440,1430,335,275,240,215,220,170,144,128]
})
fig1 = px.choropleth(world_data, locations='Code', locationmode='ISO-3',  # choropleth = colour each country
                     color='Population', hover_name='Country',
                     color_continuous_scale='OrRd',           # colour scale
                     title='World Population Distribution (in Millions)')
fig1.update_layout(width=950, height=550); fig1.show()
```
