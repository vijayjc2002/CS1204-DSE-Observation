## AIM

To build a cartographic (map-based) visualization of a dataset involving states in India using Python and Plotly.

## ALGORITHM

**Step 1:** Start the program.

**Step 2:** Import json, pandas, plotly.express and plotly.io.

**Step 3:** Set the Plotly renderer to 'notebook' for inline display.

**Step 4:** Load the world map outline and the India states GeoJSON from the data folder, so no internet is needed.

**Step 5:** Create a DataFrame of the literacy rate of ten Indian states.

**Step 6:** Draw an India states choropleth map using px.choropleth() with the GeoJSON of Indian states.

**Step 7:** Fit the map to India, set its size and display it.

**Step 8:** Stop the program.

## PROGRAM

_Click the code and press **Shift + Enter** to run it. The output appears just below the code._

```python
%matplotlib inline
import json
import pandas as pd
import plotly.express as px
import plotly.io as pio
from IPython.display import display, HTML
pio.renderers.default = "notebook"

# Load the map outline from the data folder (so no internet is needed)
world_topo = open('data/world_110m.json').read()
display(HTML("<script>window.PlotlyGeoAssets = window.PlotlyGeoAssets || {topojson: {}};" +
             "PlotlyGeoAssets.topojson['world_110m'] = " + world_topo + ";" + "</script>"))
india_geojson = json.load(open('data/india_state.geojson'))   # India state borders file

india_states = pd.DataFrame({
    'State': ['Kerala','Tamil Nadu','Maharashtra','Karnataka','Delhi',
              'Punjab','West Bengal','Gujarat','Odisha','Bihar'],
    'LiteracyRate': [94.0,80.1,82.3,75.4,86.2,75.8,76.3,78.0,72.9,61.8]
})
fig2 = px.choropleth(india_states, geojson=india_geojson,
                     locations='State', featureidkey='properties.NAME_1',  # match state name with map
                     color='LiteracyRate', hover_name='State',
                     color_continuous_scale='Blues',
                     title='India State-wise Literacy Rate (%)')
fig2.update_geos(fitbounds='locations', visible=False)        # zoom to India only
fig2.update_layout(width=850, height=700); fig2.show()
```
